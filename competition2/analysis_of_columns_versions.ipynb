```python
import pandas as pd
import os

def audit_notebook(rutas):
    # Diccionario para guardar columnas
    cols = {}
    
    # 1. Extraer columnas de cada versión
    for ver, path in rutas.items():
        if os.path.exists(path):
            # Leemos solo la primera fila para ser rápidos
            df_tmp = pd.read_parquet(path, engine='pyarrow')
            cols[ver] = set(df_tmp.columns)
            print(f"✅ {ver} cargada: {len(cols[ver])} columnas.")
        else:
            print(f"❌ Error: No se encuentra el archivo de {ver} en {path}")

    # 2. Comparación V1 -> V2
    if "V1" in cols and "V2" in cols:
        print(f"\n{'='*40}")
        print(f"📈 CAMBIOS: V1 ➔ V2")
        print(f"{'='*40}")
        print(f"**Añadidas:** {sorted(list(cols['V2'] - cols['V1']))}")
        print(f"\n**Eliminadas:** {sorted(list(cols['V1'] - cols['V2']))}")

    # 3. Comparación V2 -> V3
    if "V2" in cols and "V3" in cols:
        print(f"\n{'='*40}")
        print(f"📈 CAMBIOS: V2 ➔ V3")
        print(f"{'='*40}")
        print(f"**Añadidas:** {sorted(list(cols['V3'] - cols['V2']))}")
        print(f"\n**Eliminadas:** {sorted(list(cols['V2'] - cols['V3']))}")

# --- CONFIGURACIÓN DE RUTAS ---
# Ajusta estas rutas a tus archivos reales
mis_rutas_train = {
    "V1": "data3/train_ag.parquet",
    "V3": "data2/train_ag.parquet",
    "V2": "data/train_ag.parquet"
}

audit_notebook(mis_rutas_train)
```

```python
mis_rutas_test = {
    "V1": "data/test_ag.parquet",
    "V2": "data2/test_ag.parquet",
    "V3": "data3/test_ag.parquet"
}
audit_notebook(mis_rutas_test)
```
